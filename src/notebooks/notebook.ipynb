```python
# 1. Bibliotecas
import pandas as pd
import numpy as np

# 1.1 Modelo de Machine Learning, métrica de avaliação, validação e permutação
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import roc_auc_score, average_precision_score
from catboost import CatBoostClassifier
from scipy.stats import t

# 2. Carregar o dataset
ARQUIVO = "/home/joao/Downloads/Superacao/dados/MICRODADOS_ENEM_2023.csv"

df = pd.read_csv(
    ARQUIVO,
    sep=";",
    encoding="latin-1",
    low_memory=False
)

print(f"Carregadas {len(df):,} linhas do dataset original.")


# 3. Configurações do experimento:
# Nesta parte vou explicar as regras do experimento antes de começar (Elas não limpam os dados, apenas dizem como o experimento será conduzido)

# Primeiramente, vamos separar treino e teste pelo município da escola. A ideia é descobrir se o modelo funciona em cidades que ele ainda não conhece, e não apenas em alunos parecidos da mesma cidade
# A razão disso é porque em iterações passadas, o modelo aprendia um padrão baseado no tipo de município que um aluno estava, assim ocorrendo vazamento de dados (data leak) que aumentava o ROC AUC artificialmente, chegando a 0.81 
GRUPO = "CO_MUNICIPIO_PROVA"

# Estes dois municípios serão comparados na auditoria. O município da escola é usado no teste principal, já o município da prova serve p/ verificar quanta informação existe em cada opção
GRUPOS_AUDITORIA = [
    "CO_MUNICIPIO_ESC",
    "CO_MUNICIPIO_PROVA"
]

# Estas são as cinco notas usadas p/ calcular a nota média de cada participante. Os nomes são guardados numa lista p/ não precisar repetir as mesmas colunas várias vezes no código
NOTAS = [
    "NU_NOTA_CN",
    "NU_NOTA_CH",
    "NU_NOTA_LC",
    "NU_NOTA_MT",
    "NU_NOTA_REDACAO"
]

# A seed faz o sorteio dos folds acontecer sempre do mesmo jeito. Assim, qualquer pessoa consegue rodar o notebook e conferir os mesmos resultados
RANDOM_STATE = 42

# Dividimos os municípios em cinco partes. Em cada rodada, quatro partes treinam o modelo e uma parte é usada p/ testar em municípios não vistos
N_FOLDS_EXTERNOS = 5

# Dentro dos dados de treino, fazer mais uma divisão p/ acompanhar se o modelo está começando a exagerar. Essa validação interna serve apenas p/ decidir a hora de parar o treino (ela não é o resultado final)
N_SPLITS_VALIDACAO = 5

# Repetimos o embaralhamento de cada variável cinco vezes na análise de importância (Repetir ajuda a evitar que uma mudança de resultado aconteça apenas por acaso)
N_PERMUTACOES = 5

# 4. Limpeza e seleção dos dados
# Remoção de registros que não têm as informações necessárias p/ responder à pergunta do projeto
# Só entra na análise quem realmente fez as quatro provas objetivas, quem faltou a uma delas não tem como receber uma nota média comparável

total_original = len(df)

filtro_presenca = (
    (df["TP_PRESENCA_CN"] == 1) &
    (df["TP_PRESENCA_CH"] == 1) &
    (df["TP_PRESENCA_LC"] == 1) &
    (df["TP_PRESENCA_MT"] == 1)
)

n_presenca = filtro_presenca.sum()

print(f"\nRegistros originais: {total_original:,}")
print(f"Presentes nas 4 provas: {n_presenca:,}")
print(f"Eliminados por presença: {total_original - n_presenca:,}")
print(f"Percentual aproveitado: {n_presenca / total_original:.2%}")


# Criar a 1ª população (Depois do filtro de presença, guardamos apenas quem pode seguir p/ as próximas etapas)
df_presenca = df.loc[
    filtro_presenca
].copy()


# Conferir a renda (A faixa de renda será usada p/ comparar cada participante com pessoas na mesma faixa de renda que é divida na quantidade de salários mínimos.
filtro_renda = (
    df_presenca["Q006"].notna()
)

n_renda = filtro_renda.sum()

print(f"\nCom Q006 informado: {n_renda:,}")
print(f"Sem Q006: {len(df_presenca) - n_renda:,}")

# Conferir as notas (Verficar se as cinco notas necessárias para calcular a média estão preenchidas)
filtro_notas = (
    df_presenca[NOTAS]
    .notna()
    .all(axis=1)
)

n_notas = filtro_notas.sum()

print(f"\nCom todas as 5 notas: {n_notas:,}")
print(f"Sem alguma das 5 notas: {len(df_presenca) - n_notas:,}")


# Definir quem é elegível (presença, renda informada e notas completas)
filtro_elegivel = (
    filtro_renda &
    filtro_notas
)

df_elegiveis = df_presenca.loc[
    filtro_elegivel
].copy()

print(
    f"\nPopulação elegível antes do município: "
    f"{len(df_elegiveis):,}"
)

import gc

del df, df_presenca
del filtro_presenca, filtro_renda, filtro_notas, filtro_elegivel
gc.collect()

# 4. Seleção da população e auditoria dos municípios: Esta função recebe uma coluna de município e marca quais valores podem ser usados (válidos, vazios ou inválidos)
def auditar_municipio(
    dados,
    coluna
):

    bruto = dados[coluna]

    numerico = pd.to_numeric(
        bruto,
        errors="coerce"
    )

    invalido_nan = numerico.isna()

    invalido_menos1 = (
        numerico == -1
    )

    valido = (
        numerico.notna() &
        (numerico != -1)
    )

    return {
        "bruto": bruto,
        "numerico": numerico,
        "valido": valido,
        "nan": invalido_nan,
        "menos1": invalido_menos1
    }


# Comparar os municípios disponíveis: Ver a cobertura do município da escola e do município onde a prova foi feita
print("\n" + "AUDITORIA COMPARATIVA DOS MUNICÍPIOS")

auditorias_municipio = {}

# Fazer a mesma checagem p/ os dois tipos de município disponíveis no arquivo
for coluna in GRUPOS_AUDITORIA:

    resultado = auditar_municipio(
        df_elegiveis,
        coluna
    )

    auditorias_municipio[coluna] = resultado

    n_validos = resultado["valido"].sum()
    n_nan = resultado["nan"].sum()
    n_menos1 = resultado["menos1"].sum()

    print(f"\n--- {coluna} ---")

    print(
        f"Válido: "
        f"{n_validos:,} "
        f"({n_validos / len(df_elegiveis):.2%})"
    )

    print(
        f"NaN/não numérico: "
        f"{n_nan:,}"
    )

    print(
        f"-1: "
        f"{n_menos1:,}"
    )

    print(
        f"Inválido total: "
        f"{n_nan + n_menos1:,}"
    )

    if n_validos > 0:

        print(
            f"Municípios distintos: "
            f"{resultado['numerico'].loc[
                resultado['valido']
            ].nunique():,}"
        )


# Checando a cobertura por tipo de escola: Serve p/ entender quem fica de fora ao exigir município da escola válido.
for coluna in GRUPOS_AUDITORIA:

    print("\n")
    print("=" * 70)
    print(f"{coluna} × TIPO DE ESCOLA")
    print("=" * 70)

    valido = auditorias_municipio[
        coluna
    ]["valido"]

    tabela = (
        df_elegiveis
        .assign(
            MUNICIPIO_VALIDO=valido
        )
        .groupby("TP_ESCOLA")["MUNICIPIO_VALIDO"]
        .agg(
            Participantes="size",
            Com_Municipio="sum"
        )
    )

    tabela["Sem_Municipio"] = (
        tabela["Participantes"]
        - tabela["Com_Municipio"]
    )

    tabela["Percentual_Com_Municipio"] = (
        tabela["Com_Municipio"]
        /
        tabela["Participantes"]
    )

    print(
        tabela.to_string(
            float_format=lambda x: f"{x:.4f}"
        )
    )


# Checando a cobertura por estado: Esta tabela mostra se a disponibilidade do município muda muito entre as UFs
for coluna in GRUPOS_AUDITORIA:

    print("\n")
    print("=" * 70)
    print(f"{coluna} × UF DA ESCOLA")
    print("=" * 70)

    valido = auditorias_municipio[
        coluna
    ]["valido"]

    tabela = (
        df_elegiveis
        .assign(
            MUNICIPIO_VALIDO=valido
        )
        .groupby("SG_UF_ESC")["MUNICIPIO_VALIDO"]
        .agg(
            Participantes="size",
            Com_Municipio="sum"
        )
    )

    tabela["Sem_Municipio"] = (
        tabela["Participantes"]
        - tabela["Com_Municipio"]
    )

    tabela["Percentual_Com_Municipio"] = (
        tabela["Com_Municipio"]
        /
        tabela["Participantes"]
    )

    tabela = tabela.sort_values(
        "Percentual_Com_Municipio"
    )

    print(
        tabela.to_string(
            float_format=lambda x: f"{x:.4f}"
        )
    )


# Definindo a população final: A partir daqui, trabalhamos apenas com pessoas que têm um município da escola válido (Escolhemos o município da escola como regra principal para separar os folds)
municipio = auditorias_municipio[GRUPO]
municipio_valido = municipio["valido"]

df_validos = df_elegiveis.loc[municipio_valido].copy()

df_validos[GRUPO] = (
    municipio["numerico"]
    .loc[municipio_valido]
    .astype("int64")
)

print("\n" + "POPULAÇÃO ANALÍTICA")

print(
    f"População elegível antes do município: "
    f"{len(df_elegiveis):,}"
)

print(
    f"População analítica final: "
    f"{len(df_validos):,}"
)

print(
    f"Percentual aproveitado da população elegível: "
    f"{len(df_validos) / len(df_elegiveis):.2%}"
)

print(
    f"Municípios identificados: "
    f"{df_validos[GRUPO].nunique():,}"
)


# 5. Nota média e análise exploratória A média das cinco provas será a medida de desempenho usada p/ montar o TARGET.
df_validos["NOTA_MEDIA"] = (
    df_validos[NOTAS]
    .mean(axis=1)
)


# Boxplot por faixa de renda
# Gráfico apenas descritivo, apenas mostra como as notas se distribuem antes da modelagem
import matplotlib.pyplot as plt

# Colocando as faixas na ordem A, B, C... pro gráfico não ficar fora de ordem
ordem_renda = sorted(df_validos["Q006"].dropna().unique())
dados_boxplot = [
    df_validos.loc[df_validos["Q006"].eq(faixa), "NOTA_MEDIA"].to_numpy()
    for faixa in ordem_renda
]

fig, ax = plt.subplots(figsize=(15, 7))
boxplot = ax.boxplot(
    dados_boxplot,
    tick_labels=ordem_renda,
    patch_artist=True,
    showfliers=False,  # Os outliers ficam ocultos apenas p/ o gráfico não ficar poluído
    medianprops={"color": "#1f2937", "linewidth": 1.8},
)
for caixa in boxplot["boxes"]:
    caixa.set(facecolor="#93c5fd", edgecolor="#2563eb", alpha=0.85)

ax.set_title("Distribuição da nota média por faixa de renda — ENEM 2023")
ax.set_xlabel("Faixa de renda declarada (Q006)")
ax.set_ylabel("Nota média nas cinco provas")
ax.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()

# A tabela complementa o gráfico mostrando quantidade, média e mediana em cada faixa
resumo_renda = (
    df_validos.groupby("Q006", observed=False)["NOTA_MEDIA"]
    .agg(Participantes="size", Media="mean", Mediana="median", Desvio_Padrao="std")
    .reindex(ordem_renda)
)
display(resumo_renda.style.format({"Media": "{:.1f}", "Mediana": "{:.1f}", "Desvio_Padrao": "{:.1f}"}))


# 6. Target e seleção das variáveis
# Transformamos as letras de Q006 em números apenas p/ facilitar os agrupamentos e cálculos
mapa_renda = {
    "A": 0,
    "B": 1,
    "C": 2,
    "D": 3,
    "E": 4,
    "F": 5,
    "G": 6,
    "H": 7,
    "I": 8,
    "J": 9,
    "K": 10,
    "L": 11,
    "M": 12,
    "N": 13,
    "O": 14,
    "P": 15,
    "Q": 16
}

df_validos["RENDA_NUM"] = (
    df_validos["Q006"]
    .map(mapa_renda)
)

if df_validos["RENDA_NUM"].isna().any():

    categorias_invalidas = (
        df_validos.loc[
            df_validos["RENDA_NUM"].isna(),
            "Q006"
        ]
        .drop_duplicates()
        .tolist()
    )

    raise ValueError(
        "Existem categorias de Q006 não contempladas "
        f"no mapa de renda: {categorias_invalidas}"
    )

df_validos["RENDA_NUM"] = (
    df_validos["RENDA_NUM"]
    .astype("int8")
)


# Preparando a estratificação (Manter a distribuição das faixas de renda parecida entre os folds)
estrato = (
    df_validos["Q006"]
    .astype(str)
)


# Criação de variáveis auxiliares: 
# Estas colunas resumem informações já presentes nos microdados, como EJA e tempo desde a conclusão
# Criamos variáveis simples para resumir informações de trajetória escolar e deixar algumas respostas mais fáceis de usar pelo modelo
df_validos["ANO_CONCLUSAO_NAO_INFORMADO"] = (
    df_validos["TP_ANO_CONCLUIU"] == 0
).astype("int8")

df_validos["ANOS_DESDE_CONCLUSAO"] = np.select(
    [
        df_validos["TP_ANO_CONCLUIU"].between(1, 16),
        df_validos["TP_ANO_CONCLUIU"] == 17,
        df_validos["TP_ANO_CONCLUIU"] == 0
    ],
    [
        df_validos["TP_ANO_CONCLUIU"] - 1,
        17,
        -1
    ],
    default=-1
)

df_validos["IS_EJA"] = (
    df_validos["TP_ENSINO"] == 2
).astype("int8")

df_validos["OPCAO_INGLES"] = (
    df_validos["TP_LINGUA"] == 0
).astype("int8")

df_validos["ESCOLA_RURAL"] = (
    df_validos["TP_LOCALIZACAO_ESC"] == 2
).astype("int8")

df_validos["MIGROU_ESTADO"] = (
    df_validos["SG_UF_ESC"] !=
    df_validos["SG_UF_PROVA"]
).astype("int8")

# essa variável combina idade mais alta e conclusão do ensino médio como um indicador simples de defasagem
df_validos["DEFASAGEM_IDADE"] = (
    (df_validos["TP_FAIXA_ETARIA"] > 6) &
    (df_validos["TP_ST_CONCLUSAO"] == 1)
).astype("int8")


# Separando as variáveis do modelo
# Organizamos o que é categórico e numérico p/ o CatBoost tratar cada tipo corretamente
# O modelo básico usa poucas informações p/ termos uma referência simples de comparação
features_basicas = [
    "TP_SEXO",
    "TP_FAIXA_ETARIA",
    "TP_ESCOLA",
    "SG_UF_ESC",
    "SG_UF_PROVA"
]


# Variáveis categóricas são respostas por tipo, como sexo, UF ou tipo de escola. O CatBoost consegue tratar esses grupos sem fingir que uma categoria é maior que outra
categoricas = [
    "TP_SEXO",
    "TP_ST_CONCLUSAO",
    "TP_ANO_CONCLUIU",
    "TP_ESCOLA",
    "IN_TREINEIRO",
    "TP_DEPENDENCIA_ADM_ESC",
    "TP_LOCALIZACAO_ESC",
    "TP_SIT_FUNC_ESC",
    "SG_UF_ESC",
    "SG_UF_PROVA",
    "TP_COR_RACA"
]

# Variáveis numéricas são valores que podem ser usados como números, como faixa etária e tempo desde a conclusão
numericas = [
    "TP_FAIXA_ETARIA",
    "ANOS_DESDE_CONCLUSAO",
    "ANO_CONCLUSAO_NAO_INFORMADO",
    "IS_EJA",
    "OPCAO_INGLES",
    "ESCOLA_RURAL",
    "MIGROU_ESTADO",
    "DEFASAGEM_IDADE"
]

features_completas = (
    categoricas +
    numericas
)


# Conferindo possíveis dataleaks:
# Notas, renda, identificadores e municípios não podem entrar como features
# Estas colunas ficam proibidas porque entregariam a resposta ao modelo ou identificariam diretamente uma pessoa/local
colunas_proibidas = set(
    NOTAS +
    [
        "Q006",
        "RENDA_NUM",
        "NOTA_MEDIA",
        "CO_MUNICIPIO_ESC",
        "CO_MUNICIPIO_PROVA",
        "NU_INSCRICAO",
        "NU_NOTA_COMP1",
        "NU_NOTA_COMP2",
        "NU_NOTA_COMP3",
        "NU_NOTA_COMP4",
        "NU_NOTA_COMP5"
    ]
)

vazamentos = (
    colunas_proibidas
    .intersection(features_completas)
)

if vazamentos:

    raise RuntimeError(
        "Features proibidas detectadas: "
        f"{sorted(vazamentos)}"
    )

print("\nFeatures básicas:")
print(features_basicas)

print("\nFeatures completas:")
print(features_completas)


# Ajustando os tipos das colunas:
# Preencer valores ausentes e deixamos as variáveis no formato esperado pelo modelo
# Preencher respostas vazias p/ o modelo não falhar quando uma informação não foi declarada
for col in categoricas:

    df_validos[col] = (
        df_validos[col]
        .fillna("MISSING")
        .astype(str)
    )

for col in numericas:

    df_validos[col] = (
        pd.to_numeric(
            df_validos[col],
            errors="coerce"
        )
        .fillna(-1)
    )

# Manter apenas as colunas necessárias para o experimento
colunas_necessarias = list(dict.fromkeys(
    features_completas
    + [GRUPO, "Q006", "RENDA_NUM", "NOTA_MEDIA"]
))

df_validos = df_validos[colunas_necessarias].copy()

n_elegiveis = len(df_elegiveis)
# Liberar dados usados apenas na auditoria
del df_elegiveis, auditorias_municipio
gc.collect()

# Configuração do CatBoost:
# Os parâmetros dizem como o CatBoost deve apreender e foram escolhidos p/ buscar estabilidade, sem tentar aumentar a AUC a qualquer custo
# Manter uma velocidade de aprendizado menor p/ o modelo aprender aos poucos e reduzir o risco de decorar o treino
params_catboost = dict(
    iterations=4000,
    depth=8,
    learning_rate=0.03,
    l2_leaf_reg=15,
    random_strength=2.0,

    loss_function="Logloss",
    eval_metric="AUC",

    task_type="GPU",
    devices="0",
    gpu_ram_part=0.7, # serve pra limitar o uso da GPU (placa de vídeo), no meu caso usar minha em 100% desligava meu PC.

    random_seed=RANDOM_STATE,

    od_type="Iter",
    od_wait=150,

    verbose=0
)

# Criando uma função de resumo: calcular uma média e um intervalo exploratório p/ mostrar a variação entre os folds
# A faixa não é uma garantia, mas ajuda a enxergar se os folds deram resultados parecidos
def ic95_media(valores):

    valores = np.asarray(
        valores,
        dtype=float
    )

    n = len(valores)

    media = np.mean(valores)

    if n < 2:

        return (
            media,
            np.nan,
            np.nan
        )

    desvio = np.std(
        valores,
        ddof=1
    )

    erro = (
        desvio /
        np.sqrt(n)
    )

    margem = (
        t.ppf(
            0.975,
            df=n - 1
        )
        * erro
    )

    return (
        media,
        media - margem,
        media + margem
    )


# Criando o TARGET dentro de cada fold: 
# Esta função cria o rótulo que o modelo tenta prever quem ficou no grupo de melhor desempenho dentro da própria renda
# Os limites são calculados somente com o treino p/ não usar informação do teste antes da hora
# O treino define os cortes (validação e teste apenas recebem esses mesmos cortes)


def construir_target(treino, validacao, teste):
    media_renda = treino.groupby("RENDA_NUM")["NOTA_MEDIA"].mean()

    residuo_treino = (
        treino["NOTA_MEDIA"]
        - treino["RENDA_NUM"].map(media_renda)
    )

    p90 = residuo_treino.groupby(
        treino["RENDA_NUM"]
    ).quantile(0.90)

    for nome, dados in [
        ("treino", treino),
        ("validacao", validacao),
        ("teste", teste)
    ]:
        nota_esperada = dados["RENDA_NUM"].map(media_renda)
        limiar = dados["RENDA_NUM"].map(p90)

        if nota_esperada.isna().any() or limiar.isna().any():
            raise RuntimeError(
                f"{nome}: existe faixa de renda sem referência calculada no treino."
            )

        residuo = dados["NOTA_MEDIA"] - nota_esperada
        dados["TARGET"] = (residuo > limiar).astype("int8")

        if dados["TARGET"].nunique() < 2:
            raise RuntimeError(
                f"{nome}: target sem duas classes."
            )

    return treino, validacao, teste


# 7. Validação e treinamento dos modelos:
# O objetivo é testar se o modelo funciona em municípios que ele nunca viu durante o treino
# Esta divisão cria cinco testes diferentes sem misturar municípios entre treino e teste
# OBS: Repetir o teste em cinco partes deixa o resultado menos dependente de uma única divisão dos dados
sgkf_externo = StratifiedGroupKFold(
    n_splits=N_FOLDS_EXTERNOS,
    shuffle=True,
    random_state=RANDOM_STATE
)

resultados_folds = []

# Guardando a importância calculada pelo próprio CatBoost em cada fold p/ comparar depois
importancias_folds = []

# Também guardar a importância por permutação (Ela mede quanto a AUC diminui quando uma variável é embaralhada)
permutation_importances_folds = []

oof_indices = []
oof_y = []
oof_prob_baseline = []
oof_prob_completo = []


print("\n" + "INICIANDO VALIDAÇÃO EXTERNA POR MUNICÍPIO")

print(
    f"Unidade de agrupamento: {GRUPO}"
)


# Rodar os folds:
# Em cada rodada, treinamos em alguns municípios e avaliamos em municípios deixados de fora
# Agora repetimos o processo cinco vezes. Cada vez, um grupo diferente de municípios fica reservado pro teste
for fold, (idx_treino_val, idx_teste) in enumerate(
    sgkf_externo.split(
        df_validos,
        estrato,
        groups=df_validos[GRUPO]
    ),
    start=1
):

    print(
        f"\n>>> FOLD {fold}/{N_FOLDS_EXTERNOS}"
    )


    # Separando treino e teste:
    # Teste contém municípios que não aparecem no treino desta rodada.
    # Separar os municípios usados p/ aprender dos municípios usados p/ avaliar o resultado
    df_treino_val = (
        df_validos
        .iloc[idx_treino_val]
        .copy()
    )

    df_teste = (
        df_validos
        .iloc[idx_teste]
        .copy()
    )
    # Validação interna (ajuda o modelo a parar antes de começar a decorar o treino)
    estrato_interno = (
        df_treino_val["Q006"]
        .astype(str)
    )

    # Dentro do treino, essa parte serve p/ avisar quando o modelo deve parar de aprender
    sgkf_interno = StratifiedGroupKFold(
        n_splits=N_SPLITS_VALIDACAO,
        shuffle=True,
        random_state=RANDOM_STATE
    )

    idx_treino, idx_val = next(
        sgkf_interno.split(
            df_treino_val,
            estrato_interno,
            groups=df_treino_val[GRUPO]
        )
    )

    df_treino = (
        df_treino_val
        .iloc[idx_treino]
        .copy()
    )

    df_val = (
        df_treino_val
        .iloc[idx_val]
        .copy()
    )

    del df_treino_val
    gc.collect()

    # Conferir as linhas (Nenhum candidato pode aparecer em mais de uma parte do fold)
    assert set(
        df_treino.index
    ).isdisjoint(
        df_val.index
    )

    assert set(
        df_treino.index
    ).isdisjoint(
        df_teste.index
    )

    assert set(
        df_val.index
    ).isdisjoint(
        df_teste.index
    )


    # Conferir os municípios (garante que não existe município repetido entre treino, validação e teste)
    grupos_treino = set(
        df_treino[GRUPO]
    )

    grupos_val = set(
        df_val[GRUPO]
    )

    grupos_teste = set(
        df_teste[GRUPO]
    )

    assert grupos_treino.isdisjoint(
        grupos_val
    )

    assert grupos_treino.isdisjoint(
        grupos_teste
    )

    assert grupos_val.isdisjoint(
        grupos_teste
    )


    # Essa parte mostra o tamanho de cada parte, pra acompanhar se osfolds ficaram pareceidos
    print(
        f"Treino: "
        f"{len(df_treino):,} alunos | "
        f"{df_treino[GRUPO].nunique():,} municípios"
    )

    print(
        f"Validação: "
        f"{len(df_val):,} alunos | "
        f"{df_val[GRUPO].nunique():,} municípios"
    )

    print(
        f"Teste: "
        f"{len(df_teste):,} alunos | "
        f"{df_teste[GRUPO].nunique():,} municípios"
    )


    # Criar o TARGET deste fold (Os limites vêm apenas do treino pro teste continuar realmente desconhecido pelo modelo)
    (
        df_treino,
        df_val,
        df_teste
    ) = construir_target(
        df_treino,
        df_val,
        df_teste
    )


    # conferir  a distribuição do TARGET (É esperado que o grupo de melhor desempenho fique próximo de 10% em cada parte)
    print("\nDistribuição do target:")

    for nome, dados in [
        ("Treino", df_treino),
        ("Validação", df_val),
        ("Teste", df_teste)
    ]:

        prevalencia = dados["TARGET"].mean()

        positivos = dados["TARGET"].sum()

        print(
            f"{nome}: "
            f"{positivos:,} positivos / "
            f"{len(dados):,} "
            f"({prevalencia:.4%})"
        )

        if dados["TARGET"].nunique() < 2:

            raise RuntimeError(
                f"Fold {fold}: target sem duas classes "
                f"na {nome.lower()}."
            )


    # Separação de features e target:
    # Aqui é montado os dados que vão entrar nos modelos
    X_tr = df_treino[
        features_completas
    ]

    y_tr = df_treino[
        "TARGET"
    ]

    X_va = df_val[
        features_completas
    ]

    y_va = df_val[
        "TARGET"
    ]

    X_te = df_teste[
        features_completas
    ]

    y_te = df_teste[
        "TARGET"
    ]


    # Modelo mais simples: usa poucas variáveis e serve como ponto de comparação
    X_tr_basico = df_treino[
        features_basicas
    ]

    X_va_basico = df_val[
        features_basicas
    ]

    X_te_basico = df_teste[
        features_basicas
    ]

    # Confirmar que notas, renda e município não entraram nas features (pra não ter data leak)
    assert GRUPO not in features_completas
    assert "Q006" not in features_completas
    assert "RENDA_NUM" not in features_completas
    assert "NOTA_MEDIA" not in features_completas

    assert not any(
        nota in features_completas
        for nota in NOTAS
    )

    assert all(
        col in X_tr.columns
        for col in categoricas
    )
   
    # Indicando as colunas categóricas (CatBoost precisa saber quais colunas representam categorias)
    cat_basicas_idx = [
        X_tr_basico.columns.get_loc(c)
        for c in [
            "TP_SEXO",
            "TP_ESCOLA",
            "SG_UF_ESC",
            "SG_UF_PROVA"
        ]
    ]

    cat_completas_idx = [
        X_tr.columns.get_loc(c)
        for c in categoricas
    ]


    # Treinamento do modelo mais simples:
    # Ele mostra o quanto já conseguimos prever com o conjunto básico de variáveis.
    # Reiterando, treinando o modelo básico primeiro porque ele serve como uma comparação justa p/ o modelo completo
    modelo_baseline = CatBoostClassifier(
        **params_catboost
    )

    modelo_baseline.fit(
        X_tr_basico,
        y_tr,
        cat_features=cat_basicas_idx,
        eval_set=(
            X_va_basico,
            y_va
        ),
        use_best_model=True
    )

    prob_baseline = (
        modelo_baseline
        .predict_proba(
            X_te_basico
        )[:, 1]
    )


    # Modelo completo: usa todas as variáveis escolhidas p/ comparar com o modelo básico
    # Treinamento do modelo completo, com todas as variáveis permitidas
    modelo_completo = CatBoostClassifier(
        **params_catboost
    )

    modelo_completo.fit(
        X_tr,
        y_tr,
        cat_features=cat_completas_idx,
        eval_set=(
            X_va,
            y_va
        ),
        use_best_model=True
    )

    prob_completo = (
        modelo_completo
        .predict_proba(
            X_te
        )[:, 1]
    )


    # Cálculo de todas as métricas:
    # OBS: Foi utilizado ROC-AUC e PR-AUC porque o target tem bem menos positivos do que negativos
    # ROC-AUC mostra se o modelo costuma dar uma chance maior p/ quem realmente está no grupo de melhor desempenho
    # PR-AUC é útil porque só cerca de 10% das pessoas fazem parte desse grupo
    auc_baseline = roc_auc_score(
        y_te,
        prob_baseline
    )

    auc_completo = roc_auc_score(
        y_te,
        prob_completo
    )

    ap_baseline = average_precision_score(
        y_te,
        prob_baseline
    )

    ap_completo = average_precision_score(
        y_te,
        prob_completo
    )

    prevalencia = y_te.mean()

    # Cada candidato recebe uma previsão feita por um modelo que não a viu no treino.
    # Previsões feitas fora do treino são guardadas. Elas serão usadas no resultado geral mais confiável
    oof_indices.extend(
        df_teste.index.tolist()
    )

    oof_y.extend(
        y_te.tolist()
    )

    oof_prob_baseline.extend(
        prob_baseline.tolist()
    )

    oof_prob_completo.extend(
        prob_completo.tolist()
    )

    # Salvar o resultado do fold (Os cinco folds vão ser juntados em uma tabela final posteriormente)
    resultados_folds.append({

        "Fold":
            fold,

        "Alunos_Teste":
            len(df_teste),

        "Municipios_Teste":
            df_teste[GRUPO].nunique(),

        "Prevalencia":
            prevalencia,

        "Melhor_Iteracao":
            modelo_completo.get_best_iteration(),

        "ROC_AUC_Baseline":
            auc_baseline,

        "ROC_AUC_Completo":
            auc_completo,

        "Delta_ROC_AUC":
            auc_completo - auc_baseline,

        "PR_AUC_Baseline":
            ap_baseline,

        "PR_AUC_Completo":
            ap_completo,

        "Delta_PR_AUC":
            ap_completo - ap_baseline,

        "Lift_PR_AUC":
            ap_completo / prevalencia
    })


    # Essa é a importância que o próprio CatBoost calcula durante o treino
    importancias_folds.append(
        modelo_completo
        .get_feature_importance()
    )

    # Calculando importância por permutação:
    # Embaralhamos uma coluna por vez no teste e vemos o quanto a AUC piora
    print("\nImportância por Permutação no teste externo:")

    auc_original = auc_completo

    importancia_fold = {}

    for feature_idx, feature in enumerate(
        features_completas
    ):

        quedas_auc = []

        # O embaralhamento é repetido algumas vezes p/ importância não depender de uma única troca aleatória
        for repeticao in range(
            N_PERMUTACOES
        ):

            # Uma seed diferente é criada p/ cada combinação de fold, variável e repetição, deixando o resultado reproduzível com outros microdados do ENEM
            seed = (
                RANDOM_STATE
                + fold * 10000
                + feature_idx * 100
                + repeticao
            )

            rng = np.random.default_rng(
                seed
            )

            indices_permutados = (
                rng.permutation(
                    len(X_te)
                )
            )

            # O teste é copiado p/ mudar apenas uma variável por vez. Assim sabemos qual variável fez diferença na previsão
            # Também sçao copiados os dados de teste e trocamos somente uma coluna de lugar. Se a métrica cair, aquela coluna ajudava o modelo a fazer previsões
            X_permutado = X_te.copy()

            X_permutado[feature] = (
                X_te[feature]
                .to_numpy()[
                    indices_permutados
                ]
            )

            prob_permutada = (
                modelo_completo
                .predict_proba(
                    X_permutado
                )[:, 1]
            )

            auc_permutado = (
                roc_auc_score(
                    y_te,
                    prob_permutada
                )
            )

            queda_auc = (
                auc_original -
                auc_permutado
            )

            quedas_auc.append(
                queda_auc
            )

        media_queda = np.mean(
            quedas_auc
        )

        importancia_fold[feature] = (
            media_queda
        )

        print(
            f"  {feature:<30} "
            f"queda média = {media_queda:+.6f}"
        )

    permutation_importances_folds.append(
        importancia_fold
    )

# 8. Resultados e avaliação:
# Métricas, previsões OOF e importância das variáveis
df_resultados = pd.DataFrame(
    resultados_folds
)

print("\n" + "RESULTADOS POR FOLD")

print(
    df_resultados.to_string(
        index=False,
        float_format=lambda x: f"{x:.5f}"
    )
)

# Organização das previsões OOF:
# As previsões permitem calcular uma métrica geral sem usar previsão de treino
# Juntamos as previsões de todos os testes em uma única tabela
# Cada participante aparece uma vez, sempre como teste de um modelo que não o viu no treino
df_oof = pd.DataFrame({

    "INDEX":
        oof_indices,

    "TARGET":
        oof_y,

    "PROB_BASELINE":
        oof_prob_baseline,

    "PROB_COMPLETO":
        oof_prob_completo
})

df_oof = (
    df_oof
    .sort_values("INDEX")
    .reset_index(drop=True)
)


# Cada participante da população final deve aparecer uma única vez nas previsões OOF
assert len(df_oof) == len(df_validos)

assert df_oof["INDEX"].is_unique

print("\n" + "AUDITORIA OOF")

print(
    f"Participantes na população analítica: "
    f"{len(df_validos):,}"
)

print(
    f"Participantes com previsão OOF: "
    f"{len(df_oof):,}"
)

print(
    "Cada participante recebeu exatamente uma previsão OOF."
)


# Cálculo das métricas OOF: Este é o resumo principal, pois cada previsão foi feita fora da amostra
# Agora é calculado as métricas usando todas as previsões fora da amostra juntas
auc_oof_baseline = roc_auc_score(
    df_oof["TARGET"],
    df_oof["PROB_BASELINE"]
)

auc_oof_completo = roc_auc_score(
    df_oof["TARGET"],
    df_oof["PROB_COMPLETO"]
)

ap_oof_baseline = average_precision_score(
    df_oof["TARGET"],
    df_oof["PROB_BASELINE"]
)

ap_oof_completo = average_precision_score(
    df_oof["TARGET"],
    df_oof["PROB_COMPLETO"]
)

prevalencia_oof = (
    df_oof["TARGET"]
    .mean()
)


# Variação entre os folds
# Além da métrica geral, são calculados as médias e intervalos p/ entender a ESTABILIDADE
media_auc, li_auc, ls_auc = (
    ic95_media(
        df_resultados[
            "ROC_AUC_Completo"
        ]
    )
)

media_ap, li_ap, ls_ap = (
    ic95_media(
        df_resultados[
            "PR_AUC_Completo"
        ]
    )
)

media_delta_auc, li_delta_auc, ls_delta_auc = (
    ic95_media(
        df_resultados[
            "Delta_ROC_AUC"
        ]
    )
)

media_delta_ap, li_delta_ap, ls_delta_ap = (
    ic95_media(
        df_resultados[
            "Delta_PR_AUC"
        ]
    )
)


# Resumo final
# Aqui é comprado o modelo básico e o completo usando as métricas principais
print("\n" + "RESUMO CONSOLIDADO")

print(
    f"Prevalência média: "
    f"{df_resultados['Prevalencia'].mean():.5f}"
)

print("\n    ROC-AUC   ")

print(
    f"Baseline médio: "
    f"{df_resultados['ROC_AUC_Baseline'].mean():.5f} "
    f"± "
    f"{df_resultados['ROC_AUC_Baseline'].std():.5f}"
)

print(
    f"Completo médio: "
    f"{media_auc:.5f} "
    f"(intervalo exploratório: "
    f"{li_auc:.5f} — {ls_auc:.5f})"
)

print(
    f"Baseline OOF: "
    f"{auc_oof_baseline:.5f}"
)

print(
    f"Completo OOF: "
    f"{auc_oof_completo:.5f}"
)

print(
    f"Δ ROC-AUC médio: "
    f"{media_delta_auc:+.5f} "
    f"(intervalo exploratório: "
    f"{li_delta_auc:.5f} — {ls_delta_auc:.5f})"
)

print(
    f"Δ ROC-AUC OOF: "
    f"{auc_oof_completo - auc_oof_baseline:+.5f}"
)


print("\n    PR-AUC   ")

print(
    f"Baseline médio: "
    f"{df_resultados['PR_AUC_Baseline'].mean():.5f} "
    f"± "
    f"{df_resultados['PR_AUC_Baseline'].std():.5f}"
)

print(
    f"Completo médio: "
    f"{media_ap:.5f} "
    f"(intervalo exploratório: "
    f"{li_ap:.5f} — {ls_ap:.5f})"
)

print(
    f"Baseline OOF: "
    f"{ap_oof_baseline:.5f}"
)

print(
    f"Completo OOF: "
    f"{ap_oof_completo:.5f}"
)

print(
    f"Δ PR-AUC médio: "
    f"{media_delta_ap:+.5f} "
    f"(intervalo exploratório: "
    f"{li_delta_ap:.5f} — {ls_delta_ap:.5f})"
)

print(
    f"Δ PR-AUC OOF: "
    f"{ap_oof_completo - ap_oof_baseline:+.5f}"
)

print(
    f"\nLift PR-AUC médio: "
    f"{df_resultados['Lift_PR_AUC'].mean():.2f}x"
)

print(
    f"Prevalência OOF: "
    f"{prevalencia_oof:.5f}"
)


# Conferior se o ganho foi consistente (A ideia é ver se o modelo completo melhora em todos os folds)
print("\n" + "CONSISTÊNCIA DO GANHO DO MODELO COMPLETO")

ganhos_auc = (
    df_resultados[
        "Delta_ROC_AUC"
    ]
)

ganhos_ap = (
    df_resultados[
        "Delta_PR_AUC"
    ]
)

print(
    f"Folds com Δ ROC-AUC > 0: "
    f"{(ganhos_auc > 0).sum()}/"
    f"{len(ganhos_auc)}"
)

print(
    f"Folds com Δ PR-AUC > 0: "
    f"{(ganhos_ap > 0).sum()}/"
    f"{len(ganhos_ap)}"
)


# Tabela da importância nativa (mostra quais variáveis mais ajudaram o CatBoost nos diferentes folds)
# As importâncias dos cinco folds são agregagas para ver quais variáveis aparecem como úteis de forma mais estável
df_importancias = pd.DataFrame(
    importancias_folds,
    columns=features_completas
)

tabela_importancias = pd.DataFrame({

    "Feature":
        features_completas,

    "Importancia_Media":
        df_importancias
        .mean()
        .values,

    "Desvio_Padrao":
        df_importancias
        .std()
        .values

})

tabela_importancias["CV"] = (
    tabela_importancias["Desvio_Padrao"]
    /
    (
        tabela_importancias[
            "Importancia_Media"
        ]
        + 1e-9
    )
)

tabela_importancias = (
    tabela_importancias
    .sort_values(
        "Importancia_Media",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\n" + "TABELA 1 : IMPORTÂNCIA NATIVA DO CATBOOST")

print(
    tabela_importancias.to_string(
        index=False,
        float_format=lambda x: f"{x:.5f}"
    )
)


# Resumoda importância por permutação: Variáveis com maior queda de AUC são mais úteis p/ as previsões deste modelo
# Aqui é feito o mesmo resumo p/ a importância por permutação
df_permutation = pd.DataFrame(
    permutation_importances_folds
)

df_permutation = (
    df_permutation[
        features_completas
    ]
)

tabela_permutation = pd.DataFrame({

    "Feature":
        features_completas,

    "Queda_ROC_AUC_Media":
        df_permutation
        .mean(axis=0)
        .values,

    "Desvio_Padrao":
        df_permutation
        .std(
            axis=0,
            ddof=1
        )
        .values
})

tabela_permutation = (
    tabela_permutation
    .sort_values(
        "Queda_ROC_AUC_Media",
        ascending=False
    )
    .reset_index(drop=True)
)


print("\n" + "TABELA 2 : PERMUTATION IMPORTANCE")

print(
    "Média e desvio-padrão da queda de ROC-AUC "
    "nos 5 folds externos."
)

print(
    "Cada feature foi permutada 5 vezes por fold."
)

print(
    "Definição: AUC_original - AUC_permutado."
)

print(
    tabela_permutation.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)


# Revisão do experimento
# Este resumo registra os cuidados tomados p/ evitar dataleaks/vazamentos e avaliar o modelo.
print("\n" + "AUDITORIA FINAL")

print(
    f"População original: "
    f"{total_original:,}"
)

print(
    f"Presentes nas 4 provas: "
    f"{n_presenca:,}"
)

print(
    f"População elegível antes do município: "
    f"{n_elegiveis:,}"
)

print(
    f"População analítica final: "
    f"{len(df_validos):,}"
)

print(
    f"Unidade de agrupamento: "
    f"{GRUPO}"
)

print(
    "A unidade de agrupamento NÃO foi utilizada como feature."
)

print(
    "Q006/renda NÃO foi utilizada como feature."
)

print(
    "Notas NÃO foram utilizadas como feature."
)

print(
    "O target foi construído separadamente dentro de cada fold."
)

print(
    "A média de nota por renda utilizada no target "
    "veio somente do treino de cada fold."
)

print(
    "O P90 do residual utilizado no target "
    "veio somente do treino de cada fold."
)

print(
    "A validação interna foi utilizada exclusivamente "
    "para early stopping."
)

print(
    "O teste externo permaneceu separado do treinamento."
)

print(
    "Municípios foram mantidos integralmente fora "
    "do treino quando pertenciam ao teste."
)

print(
    "Cada participante recebeu uma previsão OOF."
)

print(
    "A comparação baseline vs. completo foi realizada "
    "nos mesmos folds."
)

print(
    "Permutation importance foi calculada somente "
    "no teste externo."
)

print(
    "Cada feature foi permutada 5 vezes em cada fold."
)

print(
    "A permutation importance usa "
    "AUC_original - AUC_permutado."
)

print(
    "O X_te original de cada fold foi preservado."
)

print(
    "O modelo não foi retreinado durante as permutações."
)

print(
    f"\nOOF ROC-AUC completo: "
    f"{auc_oof_completo:.6f}"
)

print(
    f"OOF PR-AUC completo: "
    f"{ap_oof_completo:.6f}"
)

print(
    "\nExperimento concluído"
)
```

```python
# 9. Teste de permutação: Destrói os padrões reais dos dados p/ provar que o modelo original não acertou por pura sorte (acaso)

# Rótulos são embaralhados p/ confirmar que, sem sinal real, o modelo fica perto do acaso
N_TARGET_PERMUTACOES = 5

resultados_target_perm = []

print("\n" + "Teste de Permutação")

# O teste é repetido algumas vezes p/ não tirar conclusão com base em um único embaralhamento.
for perm in range(N_TARGET_PERMUTACOES):

    print(f"\nPermutação {perm + 1}/{N_TARGET_PERMUTACOES}")

    oof_perm = np.full(
        len(df_validos),
        np.nan
    )

    for fold, (idx_treino_val, idx_teste) in enumerate(
        sgkf_externo.split(
            df_validos,
            estrato,
            groups=df_validos[GRUPO]
        ),
        start=1
    ):

        df_treino_val = (
            df_validos
            .iloc[idx_treino_val]
            .copy()
        )

        df_teste = (
            df_validos
            .iloc[idx_teste]
            .copy()
        )

        # A mesma divisão interna é repetida p/ comparar com o experimento principal
        sgkf_interno = StratifiedGroupKFold(
            n_splits=N_SPLITS_VALIDACAO,
            shuffle=True,
            random_state=RANDOM_STATE
        )

        idx_treino, idx_val = next(
            sgkf_interno.split(
                df_treino_val,
                df_treino_val["Q006"].astype(str),
                groups=df_treino_val[GRUPO]
            )
        )

        df_treino = (
            df_treino_val
            .iloc[idx_treino]
            .copy()
        )

        df_val = (
            df_treino_val
            .iloc[idx_val]
            .copy()
        )

        # O TARGET original deste fold é criado antes de embaralhar os rótulos de treino
        (
            df_treino,
            df_val,
            df_teste
        ) = construir_target(
            df_treino,
            df_val,
            df_teste
        )

        # Apenas os rótulos de treino são embaralhados p/ quebrar a relação entre features e TARGET
        rng = np.random.default_rng(
            RANDOM_STATE
            + perm * 1000
            + fold
        )

        # Os rótulos de treino são pegos e sua ordem é misturada
        # COm isso, as características deixam de ter uma ligação real com quem está no grupo de melhor desempenho
        y_tr_perm = (
            df_treino["TARGET"]
            .to_numpy()
            .copy()
        )

        rng.shuffle(y_tr_perm)

        y_tr_perm = pd.Series(
            y_tr_perm,
            index=df_treino.index
        )

        # Tabelas que entram no modelo
        X_tr = df_treino[
            features_completas
        ]

        X_va = df_val[
            features_completas
        ]

        X_te = df_teste[
            features_completas
        ]

        y_va = df_val[
            "TARGET"
        ]

        y_te = df_teste[
            "TARGET"
        ]

        # Posições das colunas categóricas são indicadas pro CatBoost
        cat_idx = [
            X_tr.columns.get_loc(c)
            for c in categoricas
        ]

        # O modelo é treinado usando os rótulos de treino embaralhados
        # O mesmo tipo de modelo é treinado, mas agora com os rótulos sem sentido
        modelo_perm = CatBoostClassifier(
            **params_catboost
        )

        modelo_perm.fit(
            X_tr,
            y_tr_perm,
            cat_features=cat_idx,
            eval_set=(X_va, y_va),
            use_best_model=True
        )

        # Previsões no teste externo, que continua com o target original
        prob_perm = (
            modelo_perm
            .predict_proba(X_te)[:, 1]
        )

        oof_perm[idx_teste] = prob_perm

    # Cálculo do AUC geral desta permutação usando todas as previsões OOF
    # OBS: Se o teste estiver funcionando, esta AUC deve ficar perto de 0.5, que representa um palpite aleatório
    auc_perm = roc_auc_score(
        df_oof["TARGET"],
        oof_perm
    )

    resultados_target_perm.append(
        auc_perm
    )

    print(
        f"AUC OOF permutado: "
        f"{auc_perm:.6f}"
    )


# AUC real é comparado com as AUCs obtidas depois de embaralhar os rótulos
media_perm = np.mean(
    resultados_target_perm
)

desvio_perm = np.std(
    resultados_target_perm,
    ddof=1
)

print("\n" + "Resultado:")

print(
    f"AUC real: "
    f"{auc_oof_completo:.6f}"
)

print(
    f"AUC permutado médio: "
    f"{media_perm:.6f}"
)

print(
    f"Desvio-padrão: "
    f"{desvio_perm:.6f}"
)

for i, auc in enumerate(
    resultados_target_perm,
    start=1
):

    print(
        f"Permutação {i}: "
        f"{auc:.6f}"
    )
```
